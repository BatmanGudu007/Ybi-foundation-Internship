# Basics of Linear Regression Model

### Simple Linear Regression Example

We will use `scikit-learn` to fit a straight line to a set of synthetic data points.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Generate random data with a linear trend
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

# Initialize and fit the model
model = LinearRegression()
model.fit(X, y)

# Print the results
print(f"Intercept (beta_0): {model.intercept_[0]:.4f}")
print(f"Slope (beta_1): {model.coef_[0][0]:.4f}")

# Predict for plotting
X_new = np.array([[0], [2]])
y_predict = model.predict(X_new)

# Plot the data and the prediction line
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='blue', label='Data points')
plt.plot(X_new, y_predict, color='red', linewidth=2, label='Linear Regression Line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Simple Linear Regression')
plt.legend()
plt.grid(True)
plt.show()
```
