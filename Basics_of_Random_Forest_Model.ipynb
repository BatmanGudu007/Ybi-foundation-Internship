# Basics of Random Forest Model

### Core Concepts of Random Forest Regression

**Random Forest** is an ensemble learning method primarily used for classification and regression tasks. It operates by constructing a multitude of decision trees during training and outputting the average prediction of the individual trees for regression tasks.

Here are the key foundational concepts:

1. **Ensemble Learning (Bagging)**
   * Random Forest uses a technique called **Bootstrap Aggregating** (or **Bagging**).
   * It trains multiple decision trees in parallel on different bootstrap samples (random samples with replacement) of the training dataset.

2. **Feature Randomness (Attribute Bagging)**
   * When splitting a node in each decision tree, the algorithm only considers a random subset of the total features rather than all available features.
   * This decorrelates the trees; if one or two features are highly dominant predictors, they won't be used at every single split across all trees, leading to a more diverse forest.

3. **Aggregation**
   * For regression, the final prediction is the average (mean) of the predictions from all the individual decision trees:
     $$\hat{y} = \frac{1}{N} \sum_{i=1}^{N} f_i(x)$$
     where $f_i(x)$ is the prediction of the $i$-th tree, and $N$ is the total number of trees.

4. **Out-of-Bag (OOB) Error**
   * Since each tree is trained on a bootstrap sample, about $1/3$ of the training data is left out (the "out-of-bag" data) for that tree. This data can be used as a built-in validation set to evaluate performance without needing a separate validation split.
