# Pandas Data Summary

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', None],
    'Age': [25, 30, 35, np.nan, 28, 40],
    'Salary': [50000, 60000, 80000, 55000, np.nan, 95000],
    'Department': ['HR', 'Engineering', 'Engineering', 'Marketing', 'HR', 'Management']
}
df = pd.DataFrame(data)

print("--- DataFrame Info ---")
df.info()

print("\n--- Basic Statistics (Numerical) ---")
print(df.describe(include=[np.number]))

print("\n--- Basic Statistics (Categorical) ---")
print(df.describe(include=[object]))

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

print("\n--- Shape of DataFrame ---")
print(df.shape)
```

```python
# Group by Department and calculate the average salary
avg_salary_by_dept = df.groupby('Department')['Salary'].mean().reset_index()

# Rename column for clarity
avg_salary_by_dept.columns = ['Department', 'Average_Salary']

display(avg_salary_by_dept)
```

```python
# Copying the original dataframe for demonstration
df_imputed = df.copy()

# Method 1: Imputation (Filling with mean or median)
# Fill missing Age with median age
df_imputed['Age'] = df_imputed['Age'].fillna(df_imputed['Age'].median())

# Fill missing Salary with mean salary
df_imputed['Salary'] = df_imputed['Salary'].fillna(df_imputed['Salary'].mean())

print("--- Dataframe after Imputation ---")
display(df_imputed)

# Method 2: Dropping rows with missing values in specific columns
df_dropped = df.dropna(subset=['Age', 'Salary'])

print("\n--- Dataframe after dropping rows with missing Age or Salary ---")
display(df_dropped)
```
