# Python Dictionary

```python
# Python Dictionary Demonstration

# 1. Creating a dictionary
person = {
    "name": "Alice",
    "age": 30,
    "city": "New York",
    "skills": ["Python", "Data Science"]
}
print("Original dictionary:", person)

# 2. Accessing elements
print("\nAccessing values:")
print("Name:", person["name"])
print("City (using get):", person.get("city"))
print("Country (with default value):", person.get("country", "Not Found"))

# 3. Modifying and adding elements
person["age"] = 31  # Update existing key
person["country"] = "USA"  # Add new key-value pair
print("\nModified dictionary:", person)

# 4. Removing elements
removed_value = person.pop("city")
print(f"\nRemoved 'city' (value: {removed_value})")
print("Dictionary after pop:", person)

# 5. Iterating through a dictionary
print("\nIterating over keys and values:")
for key, value in person.items():
    print(f"{key}: {value}")

# 6. Dictionary methods
print("\nKeys:", list(person.keys()))
print("Values:", list(person.values()))
```
