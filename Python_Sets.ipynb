# Python Sets

```python
# Creating a set
my_set = {1, 2, 3, 3, 4}  # Duplicates are automatically removed
print("Initial set:", my_set)

# Adding elements
my_set.add(5)
print("After adding 5:", my_set)

# Removing elements
my_set.remove(2)
print("After removing 2:", my_set)

# Set operations
another_set = {4, 5, 6, 7}
print("Another set:", another_set)
print("Union:", my_set.union(another_set))
print("Intersection:", my_set.intersection(another_set))
```

### Why Sets Automatically Remove Duplicates

Under the hood, Python sets are implemented using **hash tables** (similar to the keys of a Python dictionary). This implementation explains why duplicates are automatically eliminated:

1. **Mathematical Definition**: In mathematics, a set is a collection of distinct objects. Python's `set` type enforces this mathematical definition.
2. **The Hashing Mechanism**:
   * When you add an element to a set, Python calculates its **hash value** using the internal `hash()` function.
   * This hash value determines the position (or bucket) in memory where the element should be stored.
   * If you try to add another element with the same value, it will produce the same hash value and point to the same memory bucket.
3. **Value Comparison (`==`)**:
   * If a bucket is already occupied, Python checks if the existing object is equal to the new one using the `==` operator.
   * Since they are identical, Python simply discards the duplicate value rather than storing it again.

Because of this hash-table implementation, checking if an item is in a set (`element in my_set`) is extremely fast—taking $O(1)$ constant time on average, regardless of how large the set is.

### Handling Custom Objects in Sets

For an object to be stored in a Python set, it must be **hashable**. By default, user-defined custom class instances are hashable because Python uses their memory address (`id()`) as their hash value, and checks equality using `is` (identity).

However, if you want two distinct instances with the same internal data to be treated as duplicates and kept unique within a set, you must override two special methods in your class:

1. **`__eq__(self, other)`**: Defines what makes two objects "equal".
2. **`__hash__(self)`**: Generates an integer representing the object. If two objects are equal according to `__eq__`, they **must** return the same hash value.

Let's see this in action.

```python
class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id

    def __eq__(self, other):
        if not isinstance(other, Member):
            return False
        # Two members are equal if they have the same ID
        return self.member_id == other.member_id

    def __hash__(self):
        # Use the hash of the unique identifier
        return hash(self.member_id)

    def __repr__(self):
        return f"Member({self.name}, ID: {self.member_id})"

# Create a set with duplicate entries of Member
members_set = {
    Member("Alice", 101),
    Member("Bob", 102),
    Member("Alice Variant", 101) # This has duplicate ID 101
}

# Note how the duplicate member ID 101 is automatically removed
display(members_set)
```
