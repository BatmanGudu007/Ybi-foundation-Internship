# Python Tuples

```python
# Python Tuples: Creation, Accessing, and Operations

# 1. Creating a Tuple
my_tuple = ("apple", "banana", "cherry", "orange", "kiwi")
print("Original tuple:", my_tuple)
print("Type:", type(my_tuple))

# Tuples can also be created without parentheses (tuple packing)
another_tuple = 1, 2, 3
print("Packed tuple:", another_tuple)

# Single item tuple needs a trailing comma
single_item_tuple = ("apple",)
print("Single item tuple:", type(single_item_tuple))

# 2. Accessing Tuple Items
print("First item:", my_tuple[0])
print("Last item:", my_tuple[-1])
print("Range of items (index 1 to 3):", my_tuple[1:4])

# 3. Tuples are Immutable
try:
    # This will raise a TypeError because you cannot change tuple elements
    my_tuple[0] = "strawberry"
except TypeError as e:
    print("\nImmutability Demo: Cannot change tuple elements directly!")
    print("Error:", e)

# How to update a tuple (by converting to a list)
temp_list = list(my_tuple)
temp_list[0] = "strawberry"
my_tuple = tuple(temp_list)
print("Updated tuple (after list conversion):", my_tuple)

# 4. Tuple Unpacking
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits
print("\nUnpacked values:")
print("green:", green)
print("yellow:", yellow)
print("red:", red)
```
